# Independent BLS search and catalogue period checks
Run the CLI experiment first. All calculations live in the package.

```python
from pathlib import Path
import pandas as pd
import numpy as np
from tess_anomaly.config import load_config
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
cfg = load_config(ROOT / "config/small_test.yaml")
```

```python
from tess_anomaly.bls import search_bls, phase
features = pd.read_csv(ROOT / "data/features/features.csv")
with np.load(ROOT / features.iloc[0].processed_path) as archive:
    curve = dict(archive)
result, periodogram = search_bls(curve, cfg["bls"])
result
```

```python
import matplotlib.pyplot as plt
plt.plot(periodogram["period"], periodogram["power"])
plt.xlabel("Period [days]"); plt.ylabel("BLS power")
```

```python
plt.scatter(phase(curve["time"], result["best_period"], result["transit_epoch"]), curve["flux"], s=1)
plt.xlabel("Time from transit [days]"); plt.ylabel("Relative flux")
```
